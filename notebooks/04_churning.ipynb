```python
import pandas as pd

df4 = pd.read_csv('../data/BankChurners.csv')

print(df4['Attrition_Flag'].value_counts())
print(df4['Attrition_Flag'].value_counts(normalize=True))

from sklearn.model_selection import train_test_split

X = df4.drop(['Attrition_Flag', 'CLIENTNUM'], axis=1)
y = df4['Attrition_Flag']

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, stratify=y, random_state=42
)

print(y_train.value_counts(normalize=True))
print(y_test.value_counts(normalize=True))

# Gender: binary encoding
X_train['Gender'] = X_train['Gender'].map({'M': 0, 'F': 1})
X_test['Gender'] = X_test['Gender'].map({'M': 0, 'F': 1})

# Income_Category: ordinal encoding (order matters)
income_order = {
    'Less than $40K': 0,
    '$40K - $60K': 1,
    '$60K - $80K': 2,
    '$80K - $120K': 3,
    '$120K +': 4,
    'Unknown': -1
}
X_train['Income_Category'] = X_train['Income_Category'].map(income_order)
X_test['Income_Category'] = X_test['Income_Category'].map(income_order)

# Everything else non-numeric: one-hot encoding
cat_cols = ['Education_Level', 'Marital_Status', 'Card_Category']
X_train = pd.get_dummies(X_train, columns=cat_cols)
X_test = pd.get_dummies(X_test, columns=cat_cols)

# align columns in case one-hot creates mismatched columns between train/test
X_train, X_test = X_train.align(X_test, join='left', axis=1, fill_value=0)

print(X_train.dtypes)

from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix, classification_report

log_reg4 = LogisticRegression(max_iter=1000, random_state=42)
log_reg4.fit(X_train, y_train)

y_pred4 = log_reg4.predict(X_test)

print(confusion_matrix(y_test, y_pred4))
print(classification_report(y_test, y_pred4))

print(X_train.columns.tolist())

leak_cols = [
    'Naive_Bayes_Classifier_Attrition_Flag_Card_Category_Contacts_Count_12_mon_Dependent_count_Education_Level_Months_Inactive_12_mon_1',
    'Naive_Bayes_Classifier_Attrition_Flag_Card_Category_Contacts_Count_12_mon_Dependent_count_Education_Level_Months_Inactive_12_mon_2'
]
X_train = X_train.drop(columns=leak_cols)
X_test = X_test.drop(columns=leak_cols)

log_reg4 = LogisticRegression(max_iter=1000, random_state=42)
log_reg4.fit(X_train, y_train)
y_pred4 = log_reg4.predict(X_test)

print(confusion_matrix(y_test, y_pred4))
print(classification_report(y_test, y_pred4))

from sklearn.ensemble import RandomForestClassifier

rf4 = RandomForestClassifier(random_state=42)
rf4.fit(X_train, y_train)

y_pred_rf4 = rf4.predict(X_test)

print(confusion_matrix(y_test, y_pred_rf4))
print(classification_report(y_test, y_pred_rf4))
```
