```python
import pandas as pd

df2 = pd.read_csv('../data/UCI_Credit_Card.csv')
df2.info()

print(df2.duplicated().sum())
print(df2['default.payment.next.month'].value_counts())

features2 = ['LIMIT_BAL', 'SEX', 'EDUCATION', 'MARRIAGE', 'AGE',
             'PAY_0', 'PAY_2', 'PAY_3', 'PAY_4', 'PAY_5', 'PAY_6',
             'BILL_AMT1', 'BILL_AMT2', 'BILL_AMT3', 'BILL_AMT4', 'BILL_AMT5', 'BILL_AMT6',
             'PAY_AMT1', 'PAY_AMT2', 'PAY_AMT3', 'PAY_AMT4', 'PAY_AMT5', 'PAY_AMT6']

X2 = df2[features2]
y2 = df2['default.payment.next.month']

from sklearn.model_selection import train_test_split

X2_train, X2_test, y2_train, y2_test = train_test_split(
    X2, y2, test_size=0.2, random_state=42, stratify=y2
)

print(X2_train.shape, X2_test.shape)

from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler

scaler2 = StandardScaler()
X2_train_scaled = scaler2.fit_transform(X2_train)
X2_test_scaled = scaler2.transform(X2_test)

model = LogisticRegression(max_iter=1000, random_state=42)
model.fit(X2_train_scaled, y2_train)

y2_pred = model.predict(X2_test_scaled)

from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y2_test, y2_pred)
print(cm)

from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier(n_estimators=100, random_state=42, class_weight='balanced')
rf_model.fit(X2_train_scaled, y2_train)

y2_pred_rf = rf_model.predict(X2_test_scaled)

from sklearn.metrics import confusion_matrix

cm_rf = confusion_matrix(y2_test, y2_pred_rf)
print(cm_rf)
```
